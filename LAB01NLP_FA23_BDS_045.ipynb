```python
# TASK 01: Strings as Text Storage
# =====================================================================
print("=" * 60)
print("TASK 01: Strings as Text Storage")
print("=" * 60)

# (a)
sentence = "Artificial Intelligence is transforming Natural Language Processing"
print("\n(a)")
print("Sentence:", sentence)
print("Data type:", type(sentence))
print("Total characters:", len(sentence))

# (b)
print("\n(b)")
first_word = sentence[:sentence.index(" ")]      # slicing up to first space
last_char = sentence[-1]                          # negative indexing
start = sentence.index("Language")
end = start + len("Language")
lang_word = sentence[start:end]                   # specific slice range
print("First word:", first_word)
print("Last character:", last_char)
print("Extracted word:", lang_word)

# (c)
print("\n(c)")
exists = "Intelligence" in sentence
print("Does 'Intelligence' exist?", exists)
if exists:
    index_pos = sentence.find("Intelligence")
    print("Starting index of 'Intelligence':", index_pos)

# (d)
print("\n(d)")
paragraph = "NLP helps computers understand human language.\nDeep learning has improved translation quality.\nChatbots use NLP to hold conversations."
lines = paragraph.split("\n")
for line in lines:
    print(line)

```

```python
# =====================================================================
# TASK 02: Lists for Token Management
# =====================================================================
print("\n" + "=" * 60)
print("TASK 02: Lists for Token Management")
print("=" * 60)

# (a)
tokens = ["deep", "learning", "helps", "machines", "understand", "text"]
print("\n(a)")
print("Tokens:", tokens)
print("Length:", len(tokens))
print("Element at index 2:", tokens[2])
print("Last element:", tokens[-1])

# (b)
print("\n(b)")
tokens.append("efficiently")
print("After append:", tokens)
tokens.insert(3, "computers")
print("After insert:", tokens)
tokens.remove("helps")
print("After remove:", tokens)

# (c)
print("\n(c)")
tokens_A = ["NLP", "uses", "Python"]
tokens_B = ["for", "text", "analysis"]
full_tokens = tokens_A + tokens_B
print("Combined list:", full_tokens)
print("Last three elements:", full_tokens[-3:])

# (d)
print("\n(d)")
sorted_tokens = sorted(full_tokens)
length_sorted = sorted(full_tokens, key=len)
print("Alphabetically sorted:", sorted_tokens)
print("Sorted by length:", length_sorted)
# Comment: Sorting by length is often more useful in NLP because it groups
# short function words (like "for") separately from longer content words,
# which can help with tasks like stopword filtering or readability scoring.
# Alphabetical sorting is mainly useful for vocabulary indexing/lookup.
```

```python

# =====================================================================
# TASK 03: Dictionaries for Word Mapping
# =====================================================================
print("\n" + "=" * 60)
print("TASK 03: Dictionaries for Word Mapping")
print("=" * 60)

# (a)
word_freq = {"machine": 4, "learning": 7, "data": 5, "model": 3, "text": 6}
print("\n(a)")
print("Dictionary:", word_freq)
print("Frequency of 'learning':", word_freq["learning"])
word_freq["data"] = 9
print("After updating 'data':", word_freq)

# (b)
print("\n(b)")
word_freq["neural"] = 8
word_freq["network"] = 5
print("After adding entries:", word_freq)
del word_freq["model"]
print("After deleting 'model':", word_freq)
if "text" in word_freq:
    print("'text' exists in the dictionary.")
else:
    print("'text' does not exist in the dictionary.")

# (c)
print("\n(c)")
vocab = {"nlp": 0, "python": 1, "token": 2, "vector": 3, "corpus": 4}
id_to_word = {v: k for k, v in vocab.items()}
print("vocab:", vocab)
print("id_to_word:", id_to_word)

# (d)
print("\n(d)")
for word, freq in word_freq.items():
    print(f'Word: "{word}" \u2192 Frequency: {freq}')
print("Total unique words:", len(word_freq))

```

```python

# =====================================================================
# TASK 04: Sets for Unique Vocabulary
# =====================================================================
print("\n" + "=" * 60)
print("TASK 04: Sets for Unique Vocabulary")
print("=" * 60)

# (a)
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab_set = set(raw)
print("\n(a)")
print("Raw list:", raw)
print("Vocab set:", vocab_set)
print("Number of unique words:", len(vocab_set))
# Comment: Sets automatically remove duplicates because they are implemented
# as hash tables that only store one entry per unique hash value, so adding
# an element that already exists has no effect on the set's contents.

# (b)
print("\n(b)")
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}
print("Union:", set_A | set_B)
print("Intersection:", set_A & set_B)
print("Difference (A - B):", set_A - set_B)

# (c)
print("\n(c)")
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens_c = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]
cleaned_tokens = set(tokens_c) - stopwords
print("Cleaned tokens:", cleaned_tokens)

# (d)
print("\n(d)")
vocab_set.add("transformer")
vocab_set.add("bert")
vocab_set.discard("python")
print("Is 'ai' still present?", "ai" in vocab_set)
print("Final set:", vocab_set)
```

```python

# =====================================================================
# TASK 05: Tuples for Fixed Linguistic Data
# =====================================================================
print("\n" + "=" * 60)
print("TASK 05: Tuples for Fixed Linguistic Data")
print("=" * 60)

# (a)
pos_tags = [("language", "NOUN"), ("processes", "VERB"), ("human", "ADJ"),
            ("text", "NOUN"), ("quickly", "ADV")]
print("\n(a)")
print("pos_tags:", pos_tags)
print("Word of second tuple:", pos_tags[1][0])
print("Tag of second tuple:", pos_tags[1][1])

# (b)
print("\n(b)")
word, tag = pos_tags[0]
print("word:", word)
print("tag:", tag)
try:
    pos_tags[0][1] = "VERB"
except TypeError as e:
    print("Error:", e)

# (c)
print("\n(c)")
def extract_nouns(pos_list):
    return [w for w, t in pos_list if t == "NOUN"]

nouns = extract_nouns(pos_tags)
print("Nouns:", nouns)

# (d)
print("\n(d)")
pipeline_steps = ("load_data", "clean_text", "tokenize", "remove_stopwords", "vectorize")
print("Pipeline:", pipeline_steps)
print("Position of 'tokenize':", pipeline_steps.index("tokenize"))
print("Total number of steps:", len(pipeline_steps))
try:
    del pipeline_steps[0]
except TypeError as e:
    print("Error:", e)

```

```python

# =====================================================================
# TASK 06: String Cleaning Methods
# =====================================================================
print("\n" + "=" * 60)
print("TASK 06: String Cleaning Methods")
print("=" * 60)

# (a)
text = "   MACHINE Learning is AMAZING!!!   "
print("\n(a)")
stripped = text.strip()
print("After strip():", repr(stripped))
lowered = stripped.lower()
print("After lower():", lowered)
no_exclaim = lowered.replace("!", "")
print("After replace('!', ''):", no_exclaim)

# (b)
print("\n(b)")
word_list = no_exclaim.split()
print("Tokenised list:", word_list)
upper_list = [w.upper() for w in word_list]
print("Uppercase list:", upper_list)
joined = " | ".join(upper_list)
print("Joined string:", joined)

# (c)
print("\n(c)")
sentence_java = "I love Java and Java is popular"
updated_sentence = sentence_java.replace("Java", "Python")
print("Updated sentence:", updated_sentence)
print("Position of first 'Python':", updated_sentence.find("Python"))

# (d)
print("\n(d)")
def normalize(text):
    text = text.strip().lower()
    text = text.replace(",", "").replace(".", "")
    return text

print(normalize("  Hello, World.  "))
print(normalize("NLP is fun, isn't it."))
```

```python
# =====================================================================
# TASK 07: String Validation Methods
# =====================================================================
print("\n" + "=" * 60)
print("TASK 07: String Validation Methods")
print("=" * 60)

# (a)
text_nlp = "NLP is the future of AI research"
print("\n(a)")
print("Starts with 'NLP'?", text_nlp.startswith("NLP"))
print("Ends with 'research'?", text_nlp.endswith("research"))
print("Position of 'future':", text_nlp.find("future"))

# (b)
print("\n(b)")
tokens_b = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
for tok in tokens_b:
    if tok.isalpha():
        print(f"{tok}: purely alphabetic")
    elif tok.isdigit():
        print(f"{tok}: purely numeric")
    else:
        print(f"{tok}: mixed")

# (c)
print("\n(c)")
words_c = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words, pure_numbers, mixed_tokens = [], [], []
for w in words_c:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)
print("pure_words:", pure_words)
print("pure_numbers:", pure_numbers)
print("mixed_tokens:", mixed_tokens)

# (d)
print("\n(d)")
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for tok in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(tok, "->", validate_token(tok))

```

```python

# =====================================================================
# TASK 08: Functions and Lambda for NLP Pipelines
# =====================================================================
print("\n" + "=" * 60)
print("TASK 08: Functions and Lambda for NLP Pipelines")
print("=" * 60)

# (a)
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "   HELLO Natural Language Processing   "
print("\n(a)")
print("clean():", clean(sample))
print("clean_lambda():", clean_lambda(sample))

# (b)
print("\n(b)")
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

pipeline_sentence = "I love working with NLP tools in Python daily"
result_tokens = remove_short(tokenize(pipeline_sentence), 3)
print("Filtered tokens:", result_tokens)

# (c)
print("\n(c)")
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

tokens_c8 = ["language", "model", "ai", "processing", "deep"]
print("Word lengths:", word_lengths(tokens_c8))

# (d)
print("\n(d)")
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print(nlp_pipeline("  Deep Learning Models Need Lots Of Data  ", 3))
print(nlp_pipeline("  Natural Language Processing Is Fun  ", 2))

```

```python

# =====================================================================
# TASK 09: map(), filter(), sorted() and Counter
# =====================================================================
print("\n" + "=" * 60)
print("TASK 09: map(), filter(), sorted() and Counter")
print("=" * 60)

from collections import Counter

# (a)
words9 = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]
print("\n(a)")
lower_words = list(map(lambda w: w.lower(), words9))
word_lens = list(map(lambda w: len(w), words9))
print("Lowercase words:", lower_words)
print("Word lengths:", word_lens)

# (b)
print("\n(b)")
words_long = list(filter(lambda w: len(w) > 3, words9))
words_with_t = list(filter(lambda w: "t" in w.lower(), words9))
print("Words longer than 3 chars:", words_long)
print("Words containing 't':", words_with_t)

# (c)
print("\n(c)")
print("Alphabetical order:", sorted(words9))
print("By length ascending:", sorted(words9, key=len))
print("By length descending:", sorted(words9, key=len, reverse=True))

# (d)
print("\n(d)")
sentence9 = "nlp nlp python ai nlp python text ai nlp python text"
freq = Counter(sentence9.split())
print("Counter:", freq)
print("3 most common words:", freq.most_common(3))
print("Frequency of 'python':", freq["python"])
print("Total number of words:", sum(freq.values()))
print("Unique vocabulary size:", len(freq))
```
