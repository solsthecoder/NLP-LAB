```python
import pandas as pd

data = {
    'text': [
        'This movie was fantastic and thrilling from start to finish.',
        'Terrible acting and a boring plot. I hated every minute of it.',
        'An absolute masterpiece of modern cinema. Highly recommended!',
        'Waste of time. The script made no sense at all.',
        'Brilliant performances by the lead actors, a truly touching story.',
        'Dull, uninspired, and completely predictable.',
        'Loved the cinematography and the soundtrack was amazing.',
        'Not good. Slow pacing and weak character development.',
        'An incredible cinematic experience that kept me hooked.',
        'Awful direction and cringeworthy dialogue.'
    ],
    'label': [
        'positive', 'negative', 'positive', 'negative', 'positive',
        'negative', 'positive', 'negative', 'positive', 'negative'
    ]
}

df = pd.DataFrame(data)
df.to_csv('movie_reviews.csv', index=False)
print("Dataset 'movie_reviews.csv' generated successfully!")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
# Read the dataset
df = pd.read_csv('movie_reviews.csv')
# Build Bag of Words representation
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(df['text'])
# Inspect vocabulary
print("Vocabulary size:", len(vectorizer.vocabulary_))
\
# Print a small portion of the sparse matrix (first 5 documents, first 10 features)
print("\nSparse matrix portion (first 5 docs, first 10 features):")
print(X_bow[:5, :10].toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
# Build TF-IDF representation
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(df['text'])
vocab = tfidf_vectorizer.vocabulary_
# Pick two common words and a specific document (e.g., document index 1)
word1, word2 = 'the', 'was'
idx1, idx2 = vocab.get(word1), vocab.get(word2)
print(f"Document 1: '{df['text'][1]}'")
print(f"'{word1}' > Raw count: {X_bow[1, idx1]}, TF-IDF: {X_tfidf[1, idx1]:.4f}")
print(f"'{word2}' > Raw count: {X_bow[1, idx2]}, TF-IDF: {X_tfidf[1, idx2]:.4f}")

print("\nExplanation: The TF IDF weight is lower relative to its raw frequency because frequent words appearing across many documents receive a strong Inverse Document Frequency (IDF) penalty.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
# Split features and labels
X_train, X_test, y_train, y_test = train_test_split( X_tfidf, df['label'], test_size=0.2, random_state=42 )
# Train and predict with Multinomial Naive Bayes
nb_model = MultinomialNB()
nb_model.fit(X_train, y_train)
nb_preds = nb_model.predict(X_test)
# Train and predict with Logistic Regression
lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)
lr_preds = lr_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
def evaluate_and_print(model_name, y_true, y_pred):
# Calculate metrics treating 'positive' as the target class
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, pos_label='positive')
    rec = recall_score(y_true, y_pred, pos_label='positive')
    f1 = f1_score(y_true, y_pred, pos_label='positive')
    cm = confusion_matrix(y_true, y_pred, labels=['negative', 'positive'])
    print(f"--- {model_name} --")
    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    print(f"Confusion Matrix (Neg, Pos):\n{cm}\n")
evaluate_and_print("Naive Bayes", y_test, nb_preds)
evaluate_and_print("Logistic Regression", y_test, lr_preds)
```

```python
reviews_content = """An absolute waste of time and money.
Loved every single moment of this masterpiece!
The storyline was dull and predictable.
Brilliant acting and stunning visuals, highly recommended."""

with open('new_reviews.txt', 'w') as f:
    f.write(reviews_content)

print("new_reviews.txt created successfully!")
```

```python
from sklearn.pipeline import Pipeline
# Split the raw text data so the pipeline can handle vectorization internally
X_train_text, X_test_text, y_train_text, y_test_text = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42
)
# Bundling the vectorizer and classifier (using Logistic Regression as an example)
pipeline = Pipeline([
('tfidf', TfidfVectorizer()),
('clf', LogisticRegression())
])
# Train the pipeline on raw text
pipeline.fit(X_train_text, y_train_text)
# Read new reviews from the text file (assumed to exist)
try:
    with open('new_reviews.txt', 'r') as f:
        new_reviews = [line.strip() for line in f if line.strip()]
    # Classify the unseen text
    new_preds = pipeline.predict(new_reviews)
    for review, pred in zip(new_reviews, new_preds):
        print(f"Review: {review}\nPrediction: {pred}\n")
except FileNotFoundError:
    print("File 'new_reviews.txt' not found. Please ensure it exists to run predictions.")
```
