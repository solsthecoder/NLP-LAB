```python
import re
import nltk
import spacy
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# ================= TASK 1: Regex Fundamentals =================
print("=" * 50, "\nTASK 1\n" + "=" * 50)

text1 = "In 2023, Ali scored 95 marks in Lahore. Sara joined in 1999 and paid $250 for 12345 items! Hello, World."

# a) All digits
print("a) Digits:", re.findall(r"\d", text1))

# b) Words starting with a capital letter
print("b) Capitalized words:", re.findall(r"\b[A-Z][a-zA-Z]*\b", text1))

# c) Exactly 4 digits (lookarounds avoid matching part of longer numbers)
print("c) 4-digit numbers:", re.findall(r"(?<!\d)\d{4}(?!\d)", text1))

# d) Words 3 to 6 characters long
print("d) Words (3-6 chars):", re.findall(r"\b\w{3,6}\b", text1))
```

```python
# ================= TASK 2: Pattern-Based Extraction =================
print("\n" + "=" * 50, "\nTASK 2\n" + "=" * 50)

para = ("Contact support at help.desk@company.com or sales_team@shop.co.pk. "
        "Call 0300-1234567 or 0345-7654321 for urgent orders. "
        "Order ORD-4501 was placed on 15/08/2024 and shipped on 20/08/2024. "
        "A discount of 45% was given, plus a 12.5% tax rebate.")

# a) Emails
print("a) Emails:", re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", para))

# b) Phone numbers 03XX-XXXXXXX
print("b) Phones:", re.findall(r"\b03\d{2}-\d{7}\b", para))

# c) Dates DD/MM/YYYY
print("c) Dates:", re.findall(r"\b(?:0[1-9]|[12]\d|3[01])/(?:0[1-9]|1[0-2])/\d{4}\b", para))

# d) Percentages
print("d) Percentages:", re.findall(r"\d+(?:\.\d+)?%", para))

```

```python
# ================= TASK 3: Word Tokenization (NLTK vs spaCy) =================
print("\n" + "=" * 50, "\nTASK 3\n" + "=" * 50)

paragraph = ("I don't think it's going to rain today. Dr. Khan said we can't cancel the trip. "
             "They're bringing snacks, and it's only a 3-hour drive! Isn't that great?")

# a) NLTK
nltk_tokens = word_tokenize(paragraph)
print("a) NLTK:", nltk_tokens)

# b) Blank spaCy pipeline
nlp_blank = spacy.blank("en")
spacy_tokens = [t.text for t in nlp_blank(paragraph)]
print("b) spaCy:", spacy_tokens)

# c) Differences:
# 1) NLTK keeps "3-hour" as ONE token; spaCy splits it into "3", "-", "hour".
# 2) NLTK returns a plain list of strings; spaCy returns a Doc of Token objects
#    with attributes (is_alpha, like_num, ...) and whitespace information.
# (Both split contractions the same way: don't -> do + n't, it's -> it + 's)

# d) Token counts
print("d) NLTK count :", len(nltk_tokens))
print("   spaCy count:", len(spacy_tokens))
```

```python
# ================= TASK 4: Token Attributes =================
print("\n" + "=" * 50, "\nTASK 4\n" + "=" * 50)

# a) Load sentence into a Doc
sentence = "Ali paid $50 and 100 rupees; email him at ali.khan@example.com today."
doc = nlp_blank(sentence)

# b) is_alpha
print("b) is_alpha")
for token in doc:
    print(f"   {token.text:<22} {token.is_alpha}")

# c) like_num and like_email
print("c) like_num / like_email")
for token in doc:
    print(f"   {token.text:<22} like_num={token.like_num!s:<6} like_email={token.like_email}")

# d) Currency tokens only
print("d) Currency tokens:", [t.text for t in doc if t.is_currency])

```

```python
# ================= TASK 5: Sentence Segmentation =================
print("\n" + "=" * 50, "\nTASK 5\n" + "=" * 50)

# a) Load trained pipeline
nlp = spacy.load("en_core_web_sm")

text5 = ("Dr. Smith met Mr. Jones at 5 p.m. on Monday. "
         "They discussed several topics, e.g. climate change and AI. "
         "It was a productive meeting.")
doc5 = nlp(text5)

# b) Split into sentences
print("b) Sentences:")
for i, sent in enumerate(doc5.sents, 1):
    print(f"   {i}. {sent.text}")

# c) Total number of sentences
print("c) Total sentences:", len(list(doc5.sents)))

# d) Observation (one line):
# The abbreviations Dr., Mr., p.m. and e.g. did NOT cause wrong sentence breaks,
# because en_core_web_sm uses its dependency parser and context to tell they are not sentence ends.
print("d) No wrong break: en_core_web_sm handled Dr., Mr., p.m., e.g. correctly.")
```
